#Sum of Numbers

```python
n = int(input("Enter the number: "))
add = 0
for i in range(1,n+1):
    add = add+i
print("The sum of given number is",add)
```

#Product of Numbers

```python
n = int(input("Enter the number: "))
multiply = 1
for i in range(1,n+1):
    multiply = multiply*i
print("The sum of given number is",multiply)
```

#Pyramid Program

```python
n=int(input("Enter the number: "))
for i in range(1, n + 1):
        for j in range(n - i):
            print(" ", end="")

        for k in range(1, 2*i):
            print("*", end="")
        print()
```

```python
eng=input("Enter english marks: ")
math=input("Enter maths marks: ")
social=input("Enter social marks: ")
sci=input("Enter science marks: ")
french=input("Enter french marks: ")
total=(float(eng)+float(math)+float(social)+float(sci)+float(french))/5
print(total)
if total>=90 and total<=100:
  print("Distinction")
elif total>80 and total<90:
  print("outstanding")
elif total>70 and total<80:
  print("first class")
elif total>60 and total<70:
  print("second class")
elif total>50 and total<60:
  print("third class")
elif total<50:
  print("fail")
else:
  print("error")
```
