#Prime Number Logic

```python
n = int(input("Enter the number: "))
flag = False
for i in range(2,n):
  if n%i==0:
    flag = True
    break
if flag:
  print("The given number is not a prime number")
else:
  print("The given number is a prime number")
```

```python
n = int(input("Enter the number: "))
flag = False
for i in range(2,n//2):
  if n%i==0:
    flag = True
    break
if flag:
  print("The given number is not a prime number")
else:
  print("The given number is a prime number")
```

#Palindrom String logic

```python
word = "malayalam"
reverse = ""
length = len(word)-1
print(length)

while length>=0:
    reverse = reverse + word[length]
    length -=1

if word == reverse:
    print("The word is palindrome")
else:
    print("The word is not palindrome")
```

#Checking the Palindrome using typecasting method

```python
a = 234
a = str(a)
rev = ""
length=len(a)-1
while length>=0:
    rev = rev+a[length]
    length-=1
if int(a)==int(rev):
    print("This is palidrome")
else:
    print("This is not palindrome")
```
